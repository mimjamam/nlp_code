```python
import pandas as pd
import string
df=pd.read_csv('Twitter Sentiments.csv')

```

```python
df.head()
```

```python
df=df.drop(columns=['id','label'], axis=1)
```

```python
df.head()
```

```python
df['clean_text']=df['tweet'].str.lower()
```

```python
df.head()
```

```python
string.punctuation
```

```python
def remove_punctuation(text):
    punctuations =string.punctuation
    return text.translate(str.maketrans('','',punctuations))
```

```python
df['clean_text']=df['clean_text'].apply(lambda x: remove_punctuation(x))
```

```python
df.head()
```

```python
from nltk.corpus import stopwords
" , ".join(stopwords.words('english'))
```

```python
STOPWORDS= set(stopwords.words('english'))

def remove_stopwords(text):
    return " ".join([word for word in str(text).split() if word not in STOPWORDS])
```

```python
df['clean_text']=df['clean_text'].apply(lambda x: remove_stopwords(x))
```

```python
df.head()
```

```python
from collections import Counter
word_count=Counter()

for text in df['clean_text']:
    for word in text.split():
        word_count[word]+=1

word_count.most_common(10)
```

```python
FREQUENT_WORDS = set(word for (word, wc) in word_count.most_common(3))


def remove_freq_words(text):
    return " ".join([word for word in str(text).split() if word not in FREQUENT_WORDS])
```

```python
df.head()
```

```python
RARE_WORDS= set(word for (word,wc) in word_count.most_common()[:-10-1:-1])
RARE_WORDS
```

```python
def remove_rare_words(text):
    return " ".join([word for word in str(text).split() if word not in RARE_WORDS])
```

```python
df['clean_text']=df['clean_text'].apply(lambda x: remove_rare_words(x))
```

```python
df.head()
```

```python
import re

def remove_spl_char(text):
    text= re.sub('[^A-Za-z0-9]+',' ',text)
    text= re.sub('\s+',' ',text)
    return text
```

```python
df['clean_text']=df['clean_text'].apply(lambda x: remove_spl_char(x))
```

```python
df.head()
```

```python
from nltk.stem.porter import PorterStemmer

ps =PorterStemmer()
```

```python
def stem_words(text):
    return " ".join([ps.stem(word) for word in text.split()])
```

```python
df['stemmed_text']=df['clean_text'].apply(lambda x: stem_words(x))
```

```python
df.head()
```

```python
from nltk import pos_tag
from nltk.corpus import wordnet
from nltk.stem import WordNetLemmatizer

lemmatizer=WordNetLemmatizer()

wordnet_map={"N":wordnet.NOUN, "V":wordnet.VERB, "R":wordnet.ADV, "J":wordnet.ADJ}
```

```python
def lemmatize_words(text):

    pos_text=pos_tag(text.split())
    return " ".join([lemmatizer.lemmatize(word, wordnet_map.get(pos[0], wordnet.NOUN)) for word, pos in pos_text])
```

```python
wordnet.NOUN
```

```python
df['lemmatized_text']=df['clean_text'].apply(lambda x: lemmatize_words(x))
```
